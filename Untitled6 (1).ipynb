```python
import geemap
import geopandas as gpd

CITY_NAME = "Chennai"
CITY_LAT, CITY_LON = 13.0827, 80.2707

# Option A: upload your own ward geojson to Colab's file panel, then:
USE_WARDS = False   # set True once you upload a real ward file
USE_WARDS = True
if USE_WARDS:
    wards = gpd.read_file('CHENNAI.geojson')
    wards = wards.rename(columns={'Ward_No': 'ward_name'})
    wards['ward_name'] = wards['ward_name'].astype(str)   # keep it string-typed for joins
  # must have 'ward_name' column
else:
    # Option B: fallback fishnet grid (~1km cells) over a 15km box around the city center
    import numpy as np
    from shapely.geometry import box
    deg = 0.01  # ~1.1km
    half_span = 0.15  # ~15km each side
    cells = []
    ward_id = 0
    for x in np.arange(CITY_LON - half_span, CITY_LON + half_span, deg):
        for y in np.arange(CITY_LAT - half_span, CITY_LAT + half_span, deg):
            cells.append({'ward_name': f'Cell_{ward_id}', 'geometry': box(x, y, x+deg, y+deg)})
            ward_id += 1
    wards = gpd.GeoDataFrame(cells, crs='EPSG:4326')

wards_ee = geemap.geopandas_to_ee(wards)
aoi = wards_ee.geometry()
print(f"{len(wards)} zones defined.")
```

```python
START_YEAR, END_YEAR = 2015, 2024

def mask_s2(img):
    scl = img.select('SCL')
    mask = scl.neq(3).And(scl.neq(8)).And(scl.neq(9)).And(scl.neq(10))
    return img.updateMask(mask).divide(10000)

def mask_l8(img):
    qa = img.select('QA_PIXEL')
    cloud_mask = qa.bitwiseAnd(1 << 3).eq(0).And(qa.bitwiseAnd(1 << 4).eq(0))
    return img.updateMask(cloud_mask)

def get_annual_composite(year, aoi):
    start, end = f'{year}-01-01', f'{year}-12-31'
    s2 = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
          .filterDate(start, end).filterBounds(aoi)
          .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 30))
          .map(mask_s2).median())
    l8 = (ee.ImageCollection('LANDSAT/LC08/C02/T1_L2')
          .filterDate(start, end).filterBounds(aoi)
          .map(mask_l8).median())
    lst = l8.select('ST_B10').multiply(0.00341802).add(149.0).subtract(273.15).rename('LST')
    ndvi = s2.normalizedDifference(['B8', 'B4']).rename('NDVI')
    ndbi = s2.normalizedDifference(['B11', 'B8']).rename('NDBI')
    return ee.Image.cat([ndvi, ndbi, lst]).set('year', year)

print("Functions ready.")
```

```python
import pandas as pd
import time

records = []
for year in range(START_YEAR, END_YEAR + 1):
    img = get_annual_composite(year, aoi)
    try:
        stats = img.reduceRegions(collection=wards_ee, reducer=ee.Reducer.mean(), scale=100).getInfo()
        for f in stats['features']:
            p = f['properties']
            records.append({'ward': p.get('ward_name'), 'year': year,
                             'NDVI': p.get('NDVI'), 'NDBI': p.get('NDBI'), 'LST': p.get('LST')})
        print(f"Year {year} done.")
    except Exception as e:
        print(f"Year {year} failed: {e}")
    time.sleep(1)

df = pd.DataFrame(records)
df.to_csv('/content/ward_timeseries.csv', index=False)
df.head()
```

```python
def get_rainfall(year, aoi):
    return (ee.ImageCollection('UCSB-CHG/CHIRPS/DAILY')
            .filterDate(f'{year}-01-01', f'{year}-12-31').filterBounds(aoi)
            .sum().rename('RAINFALL').set('year', year))

rain_records = []
for year in range(START_YEAR, END_YEAR + 1):
    img = get_rainfall(year, aoi)
    stats = img.reduceRegions(collection=wards_ee, reducer=ee.Reducer.mean(), scale=5000).getInfo()
    for f in stats['features']:
        rain_records.append({'ward': f['properties'].get('ward_name'), 'year': year,
                              'rainfall': f['properties'].get('mean')})

rain_df = pd.DataFrame(rain_records)
df = df.merge(rain_df, on=['ward', 'year'], how='left')
df.to_csv('/content/ward_timeseries_full.csv', index=False)
```

```python
import pymannkendall as mk

trend_results = []
for ward in df['ward'].unique():
    series = df[df['ward'] == ward].sort_values('year')['LST'].dropna()
    if len(series) >= 4:
        result = mk.original_test(series)
        trend_results.append({'ward': ward, 'trend': result.trend, 'p_value': result.p, 'slope': result.slope})

trend_df = pd.DataFrame(trend_results)
trend_df.to_csv('/content/trend_significance.csv', index=False)
trend_df.sort_values('slope', ascending=False).head(10)
```

```python
import ruptures as rpt
import numpy as np

cp_results = []
for ward in df['ward'].unique():
    sub = df[df['ward'] == ward].sort_values('year')
    series = sub['NDVI'].dropna().values
    years = sub['year'].values
    if len(series) >= 5:
        algo = rpt.Pelt(model="rbf").fit(series)
        bkps = algo.predict(pen=3)
        bkp_years = [years[i] for i in bkps if i < len(years)]
        cp_results.append({'ward': ward, 'changepoint_years': bkp_years})

cp_df = pd.DataFrame(cp_results)
cp_df.to_csv('/content/changepoints.csv', index=False)
cp_df.head(10)
```

```python
from sklearn.linear_model import LinearRegression

model_df = df.dropna(subset=['NDVI', 'NDBI', 'rainfall', 'LST'])
X = model_df[['NDVI', 'NDBI', 'rainfall']]
y = model_df['LST']
model = LinearRegression().fit(X, y)

print("Coefficients:", dict(zip(X.columns, model.coef_)))
print("R²:", model.score(X, y))

import json
with open('/content/regression_summary.json', 'w') as f:
    json.dump({'coefficients': dict(zip(X.columns, model.coef_)), 'r2': model.score(X, y)}, f)
```

```python
import matplotlib.pyplot as plt

top_ward = trend_df.sort_values('slope', ascending=False).iloc[0]['ward']
sub = df[df['ward'] == top_ward].sort_values('year')

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(sub['year'], sub['LST'], marker='o', label='LST (°C)')
ax2 = ax.twinx()
ax2.plot(sub['year'], sub['NDVI'], marker='s', color='green', label='NDVI')
ax.set_xlabel('Year'); ax.set_ylabel('LST (°C)'); ax2.set_ylabel('NDVI')
plt.title(f'LST vs NDVI Trend — {top_ward}')
fig.legend(loc='upper right')
plt.savefig('/content/trend_chart.png', dpi=150, bbox_inches='tight')
plt.show()
```
